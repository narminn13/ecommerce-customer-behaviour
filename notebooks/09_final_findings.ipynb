# Final Findings — E-Commerce & Customer Behaviour Analysis

## Project Overview

This project explores an e-commerce dataset to understand data quality, product characteristics, customer profiles, order patterns, customer reviews, and purchasing behaviour.

The analysis covers five related datasets: products, customers, orders, order items, and reviews. It includes data quality auditing, exploratory data analysis (EDA), relational analysis, preprocessing and feature engineering, statistical testing, and RFM-based customer segmentation.

The main objective is to identify meaningful patterns, assess data reliability, and develop practical business recommendations while acknowledging the limitations of the available data.

## 1. Key Data Quality Findings

The initial data quality audit identified several important findings:

- **Missing values:** Product comparison prices, product ratings, customer notes, and order notes contained substantial missingness. Missing product ratings were consistent with products having zero recorded reviews.
- **Customer metric inconsistencies:** The `orders_count` and `total_spent` fields in the customer dataset did not consistently match values aggregated from the orders dataset. Each metric matched for 566 of 800 customers.
- **Referential integrity:** No invalid foreign-key references were identified between the related datasets.
- **Financial consistency:** Order-item line totals and order-level totals were consistent with their respective calculations within the specified tolerance.
- **Outliers:** Some product prices were flagged by the global IQR method, but the flags disappeared when prices were assessed within categories. This suggests that category-specific price distributions should be considered before treating these observations as anomalies.
- **Order statuses and dates:** Missing shipment and delivery timestamps were observed for some orders. These patterns should be interpreted in the context of order status and the dataset's recording practices.

### Data Quality Implications

The analysis preserved the original raw data and avoided automatically deleting outliers or imputing missing values without sufficient justification. Customer-level purchasing metrics were calculated from order records for the behavioural analysis because the existing customer summary fields were not fully consistent with those records.

## 2. Main Exploratory Data Analysis Findings

### Product Analysis
- The dataset contains 2,000 products across multiple categories and 32 brands.
- Approximately 84.9% of products were marked as in stock.
- The linear relationship between product price and product rating was negligible (Pearson \(r \approx 0.018\)) and not statistically significant.

### Customer Analysis
- The customer dataset contains 800 customers across four tiers: Standard, Silver, Gold, and Platinum.
- Standard customers formed the largest tier, representing 59% of customers.
- Approximately 55.9% of customers had marketing opt-in enabled.
- Customer order counts and total spending showed a positive association in the recorded customer summary metrics. However, inconsistencies with the orders dataset limit the reliability of this particular comparison.

### Order Analysis
- The orders dataset contains 3,000 orders, of which 89.17% were marked as delivered.
- The median order total was approximately $792.37.
- Order totals varied considerably, with some high-value orders influencing the mean.
- Discount amount and order total showed a weak positive Pearson correlation (\(r \approx 0.199\)), but the Spearman correlation was not statistically significant. This result does not establish a causal relationship.

### Review Analysis
- The mean review rating was approximately 3.91 out of 5.
- Around 70.64% of reviews received four or five stars, while 14.20% received one or two stars.
- Statistical testing found no significant difference in review ratings across product categories.
- The analysis also found no statistically significant difference in helpful votes across the defined rating groups.

### Customer Behaviour Analysis
- Of 800 customers, 766 had at least one delivered order.
- The largest RFM segment was Hibernating / Low Engagement, containing 253 customers (33.03% of customers with delivered orders).
- Champions accounted for 189 customers (24.67%).
- Delivered-order frequency and cumulative monetary value had a positive Spearman correlation of 0.649.
- Hibernating customers appeared across all customer tiers, indicating that tier labels alone do not fully describe recent purchasing activity.

## 3. Business Recommendations

Based on the findings, the following actions could be considered:

### 1. Re-engage Low-Activity Customers
The Hibernating / Low Engagement segment represents the largest RFM group. The business could test targeted re-engagement campaigns, personalised product recommendations, or limited-time incentives for these customers.

### 2. Retain High-Value Active Customers
Champions account for approximately one-quarter of customers with delivered orders. Loyalty benefits, personalised offers, and early access to relevant products could help maintain their engagement.

### 3. Investigate Customer Tier Definitions
A substantial proportion of Gold and Platinum customers were classified as Hibernating / Low Engagement. The business should review how customer tiers are assigned and whether tier membership reflects recent purchasing behaviour.

### 4. Improve Customer Metric Consistency
The mismatch between customer summary fields and order-level aggregates suggests that the data pipeline or metric definitions should be investigated. Consistent definitions of order count and customer spending would improve reporting reliability.

### 5. Investigate Customer Reviews
Although most reviews were positive, a proportion received one or two stars. Reviewing low-rated feedback may help identify recurring product or service issues.

### 6. Use Category-Aware Product Monitoring
Product prices should be evaluated within relevant categories before unusual values are flagged for investigation. A single global price threshold may not be suitable for products with different price distributions.

### Recommendation Validation

These recommendations are hypotheses derived from exploratory findings. Their effectiveness should be evaluated through follow-up analysis or controlled experiments before large-scale implementation.

## 4. Project Limitations

Several limitations should be considered when interpreting the results:

- **Customer summary inconsistencies:** Recorded customer order counts and spending did not always match aggregates calculated from the orders dataset.
- **Order status interpretation:** Delivered orders were used for RFM analysis, but recorded order totals should not automatically be treated as net revenue or profit.
- **Missing values:** Some missing values were retained because their meaning could not be established confidently. Missingness may affect particular analyses.
- **Outlier detection:** Statistical outlier flags do not necessarily indicate errors, particularly when product categories have different price ranges.
- **Statistical interpretation:** Correlations describe associations, not causation. Statistical significance does not necessarily imply practical importance.
- **RFM segmentation:** Segment definitions and scores depend on the selected observation date, scoring method, and manually defined rules.
- **Dataset scope:** Findings reflect the available dataset and may not generalise to other e-commerce businesses or periods.
- **No causal experiments:** The analysis did not use controlled experiments to establish whether the proposed business recommendations would improve outcomes.

## 5. Final Conclusion

This project examined data quality, product and customer characteristics, order patterns, customer reviews, relationships between variables, and customer purchasing behaviour.

The analysis identified inconsistencies in customer summary metrics, confirmed the validity of key dataset relationships, and found several useful behavioural patterns. Most recorded orders were marked as delivered, review ratings were generally positive, and customers with more delivered orders tended to have higher cumulative delivered-order values.

RFM segmentation revealed a substantial group of customers with relatively low engagement, providing a potential focus for future retention analysis. However, the results should be interpreted in light of the data quality issues, methodological choices, and limitations documented throughout the project.

Overall, this project demonstrates a structured approach to data auditing, exploratory analysis, statistical testing, feature engineering, and customer segmentation. It also highlights the importance of validating source data and communicating analytical findings responsibly.
